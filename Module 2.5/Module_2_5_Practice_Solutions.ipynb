## Module 1 – Descriptive Statistics

```

import numpy as np
import pandas as pd
```

### Part A – Central Tendency

```

marks = np.array([50,60,65,70,75,80,85,90])
mean = marks.mean()
median = np.median(marks)
mode = pd.Series(marks).mode().tolist()
mean, median, mode
```

```

marks_outlier = np.array([50,60,65,70,75,80,85,900])
marks_outlier.mean()
```

### Temperatures – Variance & Standard Deviation

```

temps = np.array([29,31,33,33,32,31,30])
variance = temps.var()
std = temps.std()
variance, std
```

### Part B – Percentiles & IQR

```

cats = np.array([2.5,3.0,3.2,3.3,3.4,3.5,3.6,3.9,4.0,4.5])
p25 = np.percentile(cats,25)
p50 = np.percentile(cats,50)
p75 = np.percentile(cats,75)
iqr = p75 - p25
p25, p50, p75, iqr
```

### Z-Score

```

mu, sigma = 50, 10
z_65 = (65-mu)/sigma
x_from_z = mu + (-2.5)*sigma
z_65, x_from_z
```

## Module 2 – Probability

```

# Basic Probability
P_A = 40/100
P_B = 50/100
P_A_and_B = 20/100
P_A_given_B = P_A_and_B / P_B
P_A, P_B, P_A_and_B, P_A_given_B
```

### Bayes – Spam Example

```

P_S = 100/1000
P_F_given_S = 40/100
P_F_given_notS = 60/900
P_F = P_F_given_S*P_S + P_F_given_notS*(1-P_S)
P_S_given_F = (P_F_given_S*P_S)/P_F
P_S_given_F
```

### Disease Example

```

P_D = 1/10000
P_pos_given_D = 0.99
P_pos_given_notD = 0.01
P_pos = P_pos_given_D*P_D + P_pos_given_notD*(1-P_D)
P_D_given_pos = (P_pos_given_D*P_D)/P_pos
P_D_given_pos
```

## Confusion Matrix Metrics

```

TP, FN, FP, TN = 120, 30, 80, 770
total = TP+FN+FP+TN

accuracy = (TP+TN)/total
precision = TP/(TP+FP)
recall = TP/(TP+FN)
f1 = 2*precision*recall/(precision+recall)
specificity = TN/(TN+FP)
npv = TN/(TN+FN)
prevalence = (TP+FN)/total

accuracy, precision, recall, f1, specificity, npv, prevalence
```
