Encoding, Scaling, and Distance Effects

## Step 1: Create Dataset

```

import pandas as pd

df = pd.DataFrame({
    "Income": [30000, 45000, 52000, 300000, 38000],
    "Hours_Study": [1.0, 0.5, 2.2, 5.0, 0.2],
    "GPA": [3.1, 2.6, 3.4, 3.9, 2.3],
    "City": ["Dhaka", "Chattogram", "Rajshahi", "Dhaka", "Rajshahi"],
    "Education": ["High School", "Bachelor", "Master", "Bachelor", "Master"]
})

df
```

## Step 2: Preprocessing Plan
- One-hot: City
- Ordinal: Education
- Scaling: Robust (Income), Standard (Hours_Study, GPA)

```

from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler, RobustScaler
from sklearn.compose import ColumnTransformer
```

## Step 3: ColumnTransformer

```

preprocessor = ColumnTransformer([
    ("city", OneHotEncoder(sparse=False), ["City"]),
    ("edu", OrdinalEncoder(categories=[["High School", "Bachelor", "Master"]]), ["Education"]),
    ("robust", RobustScaler(), ["Income"]),
    ("standard", StandardScaler(), ["Hours_Study", "GPA"])
])

X_transformed = preprocessor.fit_transform(df)
X_transformed[:3]
```

## Step 4: Distances Before vs After Scaling

```

from scipy.spatial.distance import euclidean, cityblock

raw = df.loc[:2, ["Income", "Hours_Study"]].values

euclidean(raw[0], raw[1]), cityblock(raw[0], raw[1])
```

## Step 5: Reflection
Robust scaling reduces the dominance of income outliers and preserves meaningful distances.
